```python

import numpy as np

# Сатып алу сомалары
purchases = np.array([
    5000, 7000, 8000, 6000, 9000,
    7500, 8500, 6500, 10000, 500000
])

# Орташа мән
mean = np.mean(purchases)

# Стандартты ауытқу
std = np.std(purchases)

# Z-score есептеу
z_scores = (purchases - mean) / std

# Аномалияларды анықтау
anomalies = purchases[np.abs(z_scores) > 2]

print("Орташа мән:", mean)
print("Стандартты ауытқу:", std)
print("Z-score:", z_scores)
print("Аномалиялар:", anomalies)

# Аномалиялардың индекстерін табу
anomaly_indices = np.where(np.abs(z_scores) > 2)[0]

print("Аномалиялардың индекстері:", anomaly_indices)

# Аномалияларды алып тастау
clean_data = purchases[np.abs(z_scores) <= 2]

print("Тазаланған деректер:", clean_data)

# Тазалаудан кейінгі орташа мән
new_mean = np.mean(clean_data)

print("Тазалауға дейінгі орташа мән:", mean)
print("Тазалаудан кейінгі орташа мән:", new_mean)
```

```python

import numpy as np

# 4 аймақ, 12 ай, 30 күн бойынша сатылым деректері
sales = np.random.randint(
    1000,
    50000,
    size=(4, 12, 30)
)

print("Массив өлшемі:", sales.shape)

# 1. Бірінші аймақтың барлық деректері
print("\n1. Бірінші аймақтың барлық деректері:")
print(sales[0])

# 2. Үшінші аймақтың 5-айдағы барлық күндік сатылымдары
print("\n2. Үшінші аймақтың 5-айдағы сатылымдары:")
print(sales[2, 4, :])

# 3. Барлық аймақтардың бірінші айдағы сатылымдары
print("\n3. Барлық аймақтардың бірінші айдағы сатылымдары:")
print(sales[:, 0, :])

# 4. Бірінші және екінші аймақтардың деректері
print("\n4. Бірінші және екінші аймақтардың деректері:")
print(sales[0:2])

# 5. Әр аймақ бойынша жылдық сатылым көлемі
print("\n5. Әр аймақ бойынша жылдық сатылым:")
print(np.sum(sales, axis=(1, 2)))

# 6. Әр ай бойынша барлық аймақтардың сатылымы
print("\n6. Әр ай бойынша барлық аймақтардың сатылымы:")
print(np.sum(sales, axis=(0, 2)))

# 7. Әр аймақтың ең жоғары күндік сатылымы
print("\n7. Әр аймақтың ең жоғары күндік сатылымы:")
print(np.max(sales, axis=(1, 2)))

# 8. Әр аймақтың ең төменгі күндік сатылымы
print("\n8. Әр аймақтың ең төменгі күндік сатылымы:")
print(np.min(sales, axis=(1, 2)))
```

```python

import numpy as np

# 1 миллион транзакция құру
n = 1_000_000

np.random.seed(42)

# Клиент ID
client_id = np.random.randint(1, 90001, n)

# Өнім ID
product_id = np.random.randint(1, 5000, n)

# Сатып алу сомасы
purchase_amount = np.random.randint(1000, 50001, n)

# Тауар саны
quantity = np.random.randint(1, 11, n)

# Жеңілдік мөлшері (%)
discount = np.random.randint(0, 31, n)

# Аймақ коды
region = np.random.randint(1, 5, n)

# Барлық бағандарды бір массивке біріктіру
transactions = np.column_stack((
    client_id,
    product_id,
    purchase_amount,
    quantity,
    discount,
    region
))

# Массивтің өлшемі мен құрылымы
print("Массив өлшемі:", transactions.shape)

print("\nБағандар:")
print("Клиент ID | Өнім ID | Сатып алу сомасы | Тауар саны | Жеңілдік (%) | Аймақ коды")

print("\nАлғашқы 10 транзакция:")
print(transactions[:10])

print("\nДеректер түрі:")
print(transactions.dtype)

print("\nТранзакциялар саны:")
print(len(transactions))
```

```python
import numpy as np

# 1. Деректердің көшірмесін құру
working_data = transactions.astype(float).copy()

# 2. Қайталанатын жазбаларды модельдеу
working_data[100:110] = working_data[0:10]

# Қайталанатын жазбаларды анықтау
unique_rows, counts = np.unique(
    working_data, axis=0, return_counts=True
)

duplicates = unique_rows[counts > 1]

print("Қайталанатын жазбалар саны:", np.sum(counts - 1))
print("Қайталанатын жазбалар:")
print(duplicates)

# 3. Жетіспейтін мәндерді модельдеу
working_data[20, 2] = np.nan
working_data[21, 4] = np.nan
working_data[22, 3] = np.nan

# Жетіспейтін мәндерді анықтау
missing_values = np.isnan(working_data)

print("\nЖетіспейтін мәндер саны:", np.sum(missing_values))
print("Әр бағандағы жетіспейтін мәндер:")
print(np.sum(missing_values, axis=0))

print("\nЖетіспейтін мәндері бар жазбалар:")
print(working_data[np.any(missing_values, axis=1)])

# 4. Қате және шектен тыс мәндерді модельдеу
working_data[30, 2] = -5000
working_data[31, 3] = 0
working_data[32, 2] = 5000000

# 5. Қате мәндерді анықтау
invalid_data = working_data[
    (working_data[:, 2] < 0) |
    (working_data[:, 3] <= 0)
]

print("\nҚате мәндері бар жазбалар:")
print(invalid_data)

print("\nҚате жазбалар саны:", len(invalid_data))

# 6. Шектен тыс мәндерді анықтау
outliers = working_data[working_data[:, 2] > 50000]

print("\nШектен тыс сатып алу сомалары:")
print(outliers)

print("\nШектен тыс мәндер саны:", len(outliers))

# 7. Деректер жиынының өлшемі
print("\nДеректер жиынының өлшемі:", working_data.shape)
```

```python
# Жетіспейтін мәндерді орташа мәнмен толтыру
clean_data = working_data.copy()

for col in range(clean_data.shape[1]):
    mean_value = np.nanmean(clean_data[:, col])
    clean_data[np.isnan(clean_data[:, col]), col] = mean_value

print("Жетіспейтін мәндер саны:")
print(np.isnan(clean_data).sum())

print("\nТазаланған деректер:")
print(clean_data[:10])
```

```python

# Қайталанатын жазбаларды жою
before = len(clean_data)

clean_data = np.unique(clean_data, axis=0)

after = len(clean_data)

print("Тазалауға дейінгі жазбалар саны:", before)
print("Тазалаудан кейінгі жазбалар саны:", after)
print("Жойылған қайталанатын жазбалар саны:", before - after)

print("\nЖаңа массив өлшемі:", clean_data.shape)
```

```python

# Қажетті бағандарды бөліп алу
selected_data = clean_data[:, [0, 2, 3]]

print("Клиент ID | Сатып алу сомасы | Тауар саны")
print(selected_data[:10])

print("\nЖаңа массив өлшемі:")
print(selected_data.shape)
```

```python

# Сатып алу сомасы 30000 теңгеден жоғары транзакциялар
filtered_data = clean_data[clean_data[:, 2] > 30000]

print("Сатып алу сомасы 30000 теңгеден жоғары транзакциялар:")
print(filtered_data[:10])

print("\nСүзгіленген транзакциялар саны:")
print(len(filtered_data))

print("\nЖаңа массив өлшемі:")
print(filtered_data.shape)
```

```python

# Жалпы табысты есептеу
total_income = np.sum(clean_data[:, 2] * (1 - clean_data[:, 4] / 100))

print("Жалпы табыс:", total_income, "теңге")
```

```python

# Әр клиенттің орташа сатып алу сомасы
client_ids, inverse = np.unique(
    clean_data[:, 0], return_inverse=True
)

total_amount = np.bincount(
    inverse, weights=clean_data[:, 2]
)

purchase_count = np.bincount(inverse)

average_amount = total_amount / purchase_count

print("Клиент ID | Орташа сатып алу сомасы")

print(np.column_stack((
    client_ids, average_amount
))[:10])
```

```python

# Әр аймақтың сатылым көлемін есептеу
region_ids, inverse = np.unique(
    clean_data[:, 5], return_inverse=True
)

region_sales = np.bincount(
    inverse,
    weights=clean_data[:, 2]
)

print("Аймақ коды | Жалпы сатылым сомасы")

print(np.column_stack((
    region_ids, region_sales
)))

print("\nСатылым көлемі ең жоғары аймақ:")
print(region_ids[np.argmax(region_sales)])

print("\nЕң жоғары сатылым сомасы:")
print(np.max(region_sales))
```
